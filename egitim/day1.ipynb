Tokanizer cümleleri kelimelere parçalayıp pozisyon bilgisi ekliyor.En basit tonizer örneği boşluğa göre bölüyor.

```python
def tokenizer(text):
    return print(text.split())

text1 = "ben uğurhan Daşşdemir llm yapmayı deneyiyorum"
tokens1 = tokenizer(text1)
```

şimdi ise bir sözlük oluşturup o sözlüğe kelimeleri ekleyip tokenize ediyoruz. Normalde sözlüğü oto olulturan olgoritmalar var. 

```python
vocab = {
    "ben" :0,
    "uğurhan": 1,
    "Daşşdemir": 2,
    "llm": 3,
    "yapmayı": 4,
    "deneyiyorum": 5,
    "<unk>" : 6
}

def tokenizer2(text):
    parts = text.split()
    ids = []

    for part in parts:
        if part in vocab:
            ids.append(vocab[part])
        else:
            ids.append(vocab["<unk>"])  # Unknown token

    return ids
tokens2 = tokenizer2(text1)
print(tokens2)  # Output: [0, 1, 2, 3, 4, 5]
```

```python
revers_vocab = {}

for part, id in vocab.items():
    revers_vocab[id]= part
revers_vocab 
```

boşluk model tokenizer da boşluk olarak değilde kelime dahil vb. şekilde farklı işleniyor

```python
def detokanizer(ids):
    parts = []
    for id in ids:
        if id in revers_vocab:
            parts.append(revers_vocab[id])
        else:
            parts.append("<unk>")  # Unknown token
    return print(" ".join(parts))

tokens3 = [0, 1, 5, 3, 2, 6]
detokanizer(tokens2)  # Output: "ben uğurhan Daşşdemir llm yapmayı deneyiyorum"
detokanizer(tokens3)  
```

https://arxiv.org/pdf/2502.07057 türkçe tokizer ilgili makale okundu, alınan notlar:

1. Türkçe gibi sondan eklemeli dilleri tokanizerlar daha küçük parçalara bölme eğilimindirler. Bu yüzden contex bağlamda şişme ve ingilizceye kıyasla perfonmans kaybı olmaktadır. Bu sorun tokenizer tasarımı ve eğitim verisiyle ilgilidir.

2. Bu çalışmada 2 krtik metirk var:
   - **2.1 — Token Saflığı (Token Purity):** tokenların ne kadar anlamlı bölündüğüdür. `["ev", "ler", "imiz", "den"]` belirli saflık sağlarken `["e", "vl", "eri", "mizden"]` saflığı düşürür.
   - **2.2 — Dile Özgü Token Yüzdeleri (Language-Specific Token Percentages):** üreilen tokenların dilde bir karşılığı olup oladığına bakar.

3. Tokenizasyonda sözlük boyutu ve dilsel doğrulu arasında denge kurmak hala zor. Aşırı parçalama anlamı sulandırırken kaba parçalama ise ayrıntıları kacırabilir. Bu denge optimize edilmeli, türkçe gibi biçimsel acıdna zor diller için bu dengenin kurulmasu zır.

4. Türkçe gibi diller için tek dilli tokenizerlar ve tokenlara fayda paunları atayan ve yüksek puanlı tokenlar üzerinde eğitim yapan eSçici Dil Modelleme'yi (Selective Language Modeling, SLM) önemli teknikleridir.

5. Türkçedeki dilsel ve kavramsal yeteneklerini ölçmek için TR-MMLU dataseti var (Sınavlardan alınmış sorular).

6. **Sözlük Boyutu (Vocabulary Size):** bir tokainezerin üretebileceği kelime sayısı, alt kelime vb. — kısaca üretebileceği tokenların büyüklüğü. Büyük olursa daha büyük kelimeler yakalanır ama bellek kullanımı artar. Azalırsa anlamsal bağlantılar kopabilir.

7. **Toplam Token Sayısı (Total Token Count):** üretilen toplam token sayısı. Az üretilmesi verimlilik sağlayabilir ama detayların kacırılmasına neden olabilir.

8. **Dile Özgü Token Yüzdeleri (%TR):** bu makalede üretilen bir metirk, üretilen tokenler tek başına dilde anlamlı mı? sorusunun cevabıdır.
   `%TR = (Geçerli Benzersiz Token'lar / Benzersiz Token'lar) × 100`

9. **Saf Token Yüzdesi (%Pure):** tokenlerdan kaç tanesi kök kelime, örneğin gül kök iken çocuklar kelimesi çoğul eki aldığı için saf değilç.
   `%Pure = (Saf Benzersiz Token'lar / Benzersiz Token'lar) × 100`

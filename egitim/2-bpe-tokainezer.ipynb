Sentencepiece ile kendi tokanizer model'nin eğitilmesi

```python
import sentencepiece as spm

spm.SentencePieceTrainer.Train(
    input="data.txt",
    model_prefix="spm_tokenizer",
    vocab_size=64,
    model_type="bpe",
)
```

```python
spm_tokenizer = spm.SentencePieceProcessor(model_file="spm_tokenizer.model")

text= "merhaba ben uğur"

spm_ids = spm_tokenizer.encode(text)
spm_tokens = spm_tokenizer.encode(text, out_type=str)
spm_ids, spm_tokens
```

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace
```

```python
hf_tokanizer = Tokenizer(BPE())
hf_tokanizer.pre_tokenizer = Whitespace()
```

```python
hf_tokanizer.get_vocab_size()
```

```python
trainer = BpeTrainer(vocab_size=64, special_tokens=["<unk>"])
hf_tokanizer.train(["data.txt"], trainer)
```

```python
hf_tokanizer.get_vocab_size(), hf_tokanizer.get_vocab()
```

```python
hf_tokanizer.save("data/hf_tokanizer.json")
```
